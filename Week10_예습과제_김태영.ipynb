```python
# Okt 토큰화
from konlpy.tag import Okt
okt = Okt() # Okt 객체의 메소드를 사용할 것임
sentence = "무엇이든 상상할 수 있는 사람은 무엇이든 만들어 낼 수 있다."

nouns = okt.nouns(sentence) # 명사 추출
phrases=okt.phrases(sentence) # 구문 추출
morphs=okt.morphs(sentence) # 형태소 추출
pos=okt.pos(sentence) # 품사 태깅

print("명사 추출: ", nouns)
print("구 추출: ", phrases)
print("형태소 추출: ", morphs)
print("품사 태깅: ", pos)
```

```python
# 꼬꼬마 토큰화
from konlpy.tag import Kkma

kkma=Kkma()
sentence = "무엇이든 상상할 수 있는 사람은 무엇이든 만들어 낼 수 있다."

nouns = kkma.nouns(sentence) # 명사 추출
sentences=kkma.sentences(sentence) # 문장 추출 (구문 추출 없음)
morphs=kkma.morphs(sentence) # 형태소 추출
pos=kkma.pos(sentence) # 품사 태깅

print("명사 추출: ", nouns)
print("문장 추출: ", sentences)
print("형태소 추출: ", morphs)
print("품사 태깅: ", pos)
```

- 꼬꼬마는 명사 추출에서 '낼', '수'를 반환하지 않음
- 꼬꼬마는 '할' -> '하', 'ㄹ'로 변환함
- 품사 태깅: Okt 총 19개 품사, 꼬꼬마 총 56개 품사 태깅

sentencepiece 라이브러리를 통해 바이트 페어 인코딩을 수행하는 토크나이저 모델 학습 실습

```python
# 청와대 청원 데이터 다운로드
from Korpora import Korpora

corpus=Korpora.load("korean_petitions") # 말뭉치 불러오기
dataset=corpus.train # 학습 데이터 가져오기
petition=dataset[0] # 첫 번째 청원 데이터

# petition 속성들
print("청원 시작일 :", petition.begin)
print("청원 종료일 :", petition.end)
print("청원 동의 수 :", petition.num_agree)
print("청원 범주 :", petition.category)
print("청원 제목 :", petition.title)
print("청원 본문 :", petition.text[:30])
```

```python
# 학습 데이터세트 생성
petitions=corpus.get_all_texts() # 본문 데이터세트를 하나의 txt 파일로 저장
with open("/content/drive/MyDrive/euron/datasets/week10/corpus.txt", "w", encoding='utf-8') as f:
  for petition in petitions:
    f.write(petition+'\n')
```

```python
# 토크나이저 모델 학습
from sentencepiece import SentencePieceTrainer

SentencePieceTrainer.Train(
    "--input=/content/drive/MyDrive/euron/datasets/week10/corpus.txt\
    --model_prefix=petition_bpe\
    --vocab_size=8000 model_type=bpe"
    # input: 말뭉치 텍스트 파일 경로
    # model_prefix: 모델 파일 이름
    # vocab_size: 어휘 사전 크기
    # model_type: 토크나이저 알고리즘 (unigram, bpe, char, word)
)
```

학습 결과
- petition_bpe.model: 학습된 토크나이저가 저장된 파일
- petition_bpe.vocab: 어휘 사전이 저장된 파일

```python
# 바이트 페어 인코딩 토큰화
from sentencepiece import SentencePieceProcessor

tokenizer = SentencePieceProcessor()
tokenizer.load("petition_bpe.model") # 학습된 모델을 불러옴

sentence = "안녕하세요, 토크나이저가 잘 학습되었군요!"
sentences = ["이렇게 입력값을 리스트로 받아서", "쉽게 토크나이저를 사용할 수 있답니다"]

# encode_as_pieces: 문장 토큰화
tokenized_sentence = tokenizer.encode_as_pieces(sentence)
tokenized_sentences = tokenizer.encode_as_pieces(sentences)
print("단일 문장 토큰화 :", tokenized_sentence)
print("여러 문장 토큰화 :", tokenized_sentences)

# encode_as_ids: 토큰을 정수형 ID와 매핑, ID 제공
encoded_sentence = tokenizer.encode_as_ids(sentence)
encoded_sentences = tokenizer.encode_as_ids(sentences)
print("단일 문장 정수 인코딩 :", encoded_sentence)
print("여러 문장 정수 인코딩 :", encoded_sentences)

# id 정수 -> 문자열 데이터 변환
decode_ids = tokenizer.decode_ids(encoded_sentences)
decode_pieces = tokenizer.decode_pieces(encoded_sentences)
print("정수 인코딩에서 문장 변환 :", decode_ids)
print("하위 단어 토큰에서 문장 변환 :", decode_pieces)
```

```python
# 어휘 사전 불러오기
vocab = {idx: tokenizer.id_to_piece(idx) for idx in range(tokenizer.get_piece_size())}
## get_piece_size: 센텐스피스 모델에서 생성된 하위 단어의 개수 반환
print(list(vocab.items())[:5]) # 하위 단어 5개 출력
print("vocab size: ", len(vocab))
```

## 워드피스 (Wordpiece)

```python
# 워드피스 토크나이저 학습
from tokenizers import Tokenizer
from tokenizers.models import WordPiece
from tokenizers.normalizers import Sequence, NFD, Lowercase
from tokenizers.pre_tokenizers import Whitespace

tokenizer = Tokenizer(WordPiece()) # 토크나이저 객체
tokenizer.normalizer=Sequence([NFD(), Lowercase()]) # 정규화 방식
## Sequence: 시퀀스 형식으로 인스턴스 전달
## 적용한 정규화 방식: NFD 유니코드 정규화, 소문자 변환(lowercase)
tokenizer.pre_tokenizer=Whitespace() # 사전 토큰화 방식 - 공백, 구두점 기준

tokenizer.train(["/content/drive/MyDrive/euron/datasets/week10/corpus.txt"]) # 데이터세트 경로 전달
tokenizer.save("/content/drive/MyDrive/euron/datasets/week10/petition_wordpiece.json") # 학습 결과 저장
## 학습 결과 -> json 파일 (메타데이터+어휘 사전)
```

```python
# 워드피스 토큰화
from tokenizers.decoders import WordPiece as WordPieceDecoder

tokenizer=Tokenizer.from_file("/content/drive/MyDrive/euron/datasets/week10/petition_wordpiece.json")
tokenizer.decoder=WordPieceDecoder() # 토크나이저의 디코더를 워드피스 디코더로 설정

sentence = "안녕하세요, 토크나이저가 잘 학습되었군요!"
sentences = ["이렇게 입력값을 리스트로 받아서", "쉽게 토크나이저를 사용할 수 있답니다"]

encoded_sentence=tokenizer.encode(sentence) # 문장 토큰화
encoded_sentences=tokenizer.encode_batch(sentences) # 여러 문장 한 번에 토큰화
## -> 문장 토큰화한 뒤 각 토큰의 색인 번호 반환

print("인코더 형식 :", type(encoded_sentence))

# tokens 속성 -> 토큰화된 데이터 값
print("단일 문장 토큰화 :", encoded_sentence.tokens)
print("여러 문장 토큰화 :", [enc.tokens for enc in encoded_sentences])

# 인코딩된 문장의 id 값
print("단일 문장 정수 인코딩 :", encoded_sentence.ids)
print("여러 문장 정수 인코딩 :", [enc.ids for enc in encoded_sentences])

# 디코딩
print("정수 인코딩에서 문장 변환 :", tokenizer.decode(encoded_sentence.ids))
```

### N-gram

```python
# N-gram 구현
import nltk

def ngrams(sentence, n):
  words=sentence.split()
  ngrams=zip(*[words[i:] for i in range(n)])
  return list(ngrams)

sentence = "안녕하세요 만나서 진심으로 반가워요"

unigram = ngrams(sentence, 1)
bigram = ngrams(sentence, 2)
trigram = ngrams(sentence, 3)

print(unigram)
print(bigram)
print(trigram)

unigram = nltk.ngrams(sentence.split(), 1)
bigram = nltk.ngrams(sentence.split(), 2)
trigram = nltk.ngrams(sentence.split(), 3)

print(list(unigram))
print(list(bigram))
print(list(trigram))
```
