# **U-Net 모델 구현**

✅ 아래 코드는 PyTorch로 작성한 간단한 U-Net 모델입니다.

✅ 주요 목표: 작은 이미지 세트로 segmentation 학습을 테스트하기!

✅ 특징: 업샘플링(deconv/upsample 선택 가능), 입력 스케일링, Gaussian noise 추가, 배치 정규화 적용

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset

# --- 작은 U-Net 정의 ---
class UNetSmall(nn.Module):
    def __init__(self, upsample_mode='deconv', net_scaling=None, edge_crop=0, gaussian_noise_std=0.1):
        super().__init__()
        self.net_scaling = net_scaling
        self.edge_crop = edge_crop
        self.gaussian_noise_std = gaussian_noise_std
        self.upsample_mode = upsample_mode

         # Convolution Block: Conv -> ReLU -> Conv -> ReLU
        def conv_block(in_c, out_c):
            return nn.Sequential(
                nn.Conv2d(in_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True),
                nn.Conv2d(out_c, out_c, 3, padding=1),
                nn.ReLU(inplace=True)
            )

        # 인코더 부분 목표 -> 128 까지!
        self.enc1 = conv_block(3, 16)
        self.enc2 = conv_block(16, 32)
        self.enc3 = conv_block(32, 64)
        self.enc4 = conv_block(64, 128)
        # Bottleneck
        self.center = conv_block(128, 256)
        self.pool = nn.MaxPool2d(2)


        # 업샘플링 부분 정의 (deconv 또는 upsample+conv) 로
        if upsample_mode == 'deconv':
            def upsample_layer(in_c, out_c):
                return nn.ConvTranspose2d(in_c, out_c, 2, stride=2)
        else:
            def upsample_layer(in_c, out_c):
                return nn.Sequential(
                    nn.Upsample(scale_factor=2, mode='nearest'),
                    nn.Conv2d(in_c, out_c, 1)
                )
        # 디코더 부분
        self.up4 = upsample_layer(256, 128)
        self.dec4 = conv_block(256, 128)
        self.up3 = upsample_layer(128, 64)
        self.dec3 = conv_block(128, 64)
        self.up2 = upsample_layer(64, 32)
        self.dec2 = conv_block(64, 32)
        self.up1 = upsample_layer(32, 16)
        self.dec1 = conv_block(32, 16)

        # 최종 출력: 1채널(segmentation mask)
        self.final = nn.Conv2d(16, 1, 1)

    def forward(self, x):
        if self.net_scaling:
        # 입력 스케일링
            x = F.avg_pool2d(x, self.net_scaling)
        # Gaussian noise 추가
        if self.gaussian_noise_std:
            noise = torch.randn_like(x) * self.gaussian_noise_std
            x = x + noise

        # 배치 정규화 적용
        x = F.batch_norm(x, running_mean=None, running_var=None, training=True)

        # 인코더
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        e4 = self.enc4(self.pool(e3))
        # Bottleneck
        center = self.center(self.pool(e4))

        # 디코더 + Skip Connection
        d4 = self.up4(center)
        d4 = torch.cat((d4, e4), dim=1)
        d4 = self.dec4(d4)

        d3 = self.up3(d4)
        d3 = torch.cat((d3, e3), dim=1)
        d3 = self.dec3(d3)

        d2 = self.up2(d3)
        d2 = torch.cat((d2, e2), dim=1)
        d2 = self.dec2(d2)

        d1 = self.up1(d2)
        d1 = torch.cat((d1, e1), dim=1)
        d1 = self.dec1(d1)

        out = self.final(d1)
        out = torch.sigmoid(out)
         # Crop + Pad
        if self.edge_crop > 0:
            out = out[:, :, self.edge_crop:-self.edge_crop, self.edge_crop:-self.edge_crop]
            out = F.pad(out, (self.edge_crop, self.edge_crop, self.edge_crop, self.edge_crop))
        # Upsample 다시 확대
        if self.net_scaling:
            out = F.interpolate(out, scale_factor=self.net_scaling, mode='nearest')

        return out
```

## **학습용 데이터 생성하기**

```python
batch_size = 32
x = torch.randn(batch_size, 3, 128, 128)  # RGB 이미지
y = (torch.rand(batch_size, 1, 128, 128) > 0.5).float()  # 랜덤 이진 마스크

dataset = TensorDataset(x, y)
loader = DataLoader(dataset, batch_size=2, shuffle=True)
```

## **데이터 시각화**

```python
import matplotlib.pyplot as plt
# 첫 번째 이미지를 시각화
sample_x = x[0]
sample_y = y[0]

plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.imshow(sample_x.permute(1, 2, 0).numpy())
plt.title('Input Image')

plt.subplot(1, 2, 2)
plt.imshow(sample_y.squeeze().numpy(), cmap='gray')
plt.title('Ground Truth Mask')
plt.show()
```

## **모델 선언 및 학습**

```python
# --- 모델 선언 및 학습 ---
model = UNetSmall(upsample_mode='deconv', net_scaling=None, edge_crop=0)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.BCELoss()

# 간단히 30 epoch만
for epoch in range(10):
    model.train()
    total_loss = 0
    for xb, yb in loader:
        pred = model(xb)
        loss = criterion(pred, yb)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    print(f"Epoch {epoch+1}, Loss: {total_loss/len(loader):.4f}")
```

[답변]

U-Net은 이미지 분할 작업을 위해 특별히 설계된 컨볼루션 신경망 아키텍처입니다.

* U자형 구조: 네트워크가 인코더(contracting path)와 디코더(expanding path)로 구성되어 U자 모양을 이룹니다.
* 인코더: 일반적인 CNN처럼 컨볼루션 및 풀링 연산을 통해 특징 맵의 해상도를 줄이면서 채널 수를 늘립니다. 이는 이미지의 맥락적 정보를 포착하는 데 도움을 줍니다.
*   디코더: 업샘플링(Upsampling) 및 컨볼루션 연산을 통해 특징 맵의 해상도를 원래 이미지 크기로 복원합니다.
*   Skip Connection: 인코더의 각 단계에서 생성된 특징 맵이 디코더의 해당하는 단계와 연결됩니다. 이는 인코더에서 손실될 수 있는 공간 정보를 디코더에 전달하여 더 정확한 경계 예측을 가능하게 합니다.


## **아래 주석을 풀고 DeepLabv1, v2, v3. v3+  모델에 대해 각각 간단히 설명해주세요 🐶**

[답변]

[DeepLab v1]
DeepLab v1은 Semantic Segmentation을 위해 제안된 초기 모델입니다. 주요 특징은 다음과 같습니다.
*   **Atrous Convolution (Dilated Convolution)**: 컨볼루션 필터의 receptive field를 확장하면서 해상도 손실 없이 더 넓은 맥락 정보를 통합할 수 있도록 합니다.
*   **Conditional Random Field (CRF)**: 초기 예측 결과를 개선하기 위해 후처리 단계로 사용되어 경계 부분을 더 정교하게 만듭니다.

[DeepLab v2]
DeepLab v2는 v1을 개선한 버전입니다.
*   **Atrous Spatial Pyramid Pooling (ASPP)**: 다양한 rate의 atrous convolution을 병렬로 적용하여 다양한 스케일의 맥락 정보를 효과적으로 캡처합니다. 이는 multi-scale 정보를 처리하는 데 강점을 가집니다.

[DeepLab v3]
DeepLab v3는 v2에서 발전하여 ASPP 모듈을 개선하고 인코더-디코더 구조를 도입했습니다.
*   **Improved ASPP**: Batch Normalization을 ASPP에 통합하고, 이미지 수준 피처를 추가하여 성능을 향상시켰습니다.
*   **Encoder-Decoder Structure**: Semantic Segmentation 태스크에 더 적합한 형태로 네트워크 구조를 개선했습니다.

[DeepLab v3+]
DeepLab v3+는 DeepLab v3의 확장 버전으로, Encoder-Decoder 구조를 더욱 강화했습니다.
*   **Decoder Module**: 인코더에서 얻은 저수준 특징(low-level features)을 활용하는 강력한 디코더를 추가하여 경계 정보를 더 잘 복원합니다.
*   **Atrous Separable Convolution**: Atrous Convolution과 Depthwise Separable Convolution을 결합하여 계산 효율성을 높였습니다.
